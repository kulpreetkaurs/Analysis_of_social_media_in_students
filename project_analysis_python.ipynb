```python
# as the data is clean and its dtypes are also accurately classified, so we now move forward to Exploratory data anylysis
columns=final_df.columns
categorical_features=[feature for feature in columns if final_df[feature].dtype=="O"]
numerical_features=[feature for feature in columns if final_df[feature].dtype!="O"]
```

```python
# Categorical features>> frequency plot, bar chart, pie chart
# Numerical features>> histogram, distplot, box plot, line chart, scatter plot1
```

```python
# categorical feature analysis
for col in categorical_features:
    print(f"{col} : {final_df[col].value_counts(normalize=True)*100}\n ")
```

```python
coun=final_df['Country'].value_counts()
plt.figure(figsize=(12,12))
plt.pie(coun,labels=coun.index,autopct="%1.1f%%",shadow=True)
plt.show()
```

```python
platform=final_df['Most_Used_Platform'].value_counts()
plt.figure(figsize=(10,12))
plt.bar(platform.index,platform,color=sns.color_palette("viridis", len(platform)))
for i, v in enumerate(platform):
    plt.text(i, v + 10, str(v), ha='center', va='bottom', fontsize=10, rotation=90)
plt.xlabel("Platfrom name")
plt.ylabel("Number of users")
plt.title("Most used Social media platform")
plt.xticks(rotation = 90)
plt.tight_layout()
plt.show()
```

```python
# most number of users are that of instagram
```

```python
acdp=final_df.Affects_Academic_Performance.value_counts()
plt.pie(acdp,labels=acdp.index,autopct="%1.1f%%",shadow=True)
plt.show()
```
